```
# ==========================================
# PEKAN 1: MEMBACA DATASET DARI FILE EXCEL BARU
# ==========================================

import pandas as pd
import matplotlib.pyplot as plt

# 1. Membaca file Excel Saham BBCA baru yang Anda unggah
file_path = "Data_Saham_BBCA.xlsx"
df = pd.read_excel(file_path)

# 2. Mengatur kolom Tanggal sebagai Index
df['Tanggal'] = pd.to_datetime(df['Tanggal'])
df.set_index('Tanggal', inplace=True)

# 3. MENYESUAIKAN NAMA KOLOM (Mengubah 'Harga Saham BBCA (IDR)' menjadi 'Harga')
df.columns = ['Harga']

# Tampilkan pengecekan data
print("=== 5 Data Pertama Saham BBCA ===")
print(df.head())
print(f"\nTotal baris data dalam file: {len(df)} baris.\n")

# 4. Visualisasi Grafik Historis Awal
plt.figure(figsize=(12, 5))
df['Harga'].plot(color='blue', linewidth=2)
plt.title('Grafik Harga Penutupan Saham BCA (BBCA) - Data Pekan 1', fontsize=14)
plt.xlabel('Tanggal', fontsize=12)
plt.ylabel('Harga (IDR)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```
# =====================================================================
# GABUNGAN PEKAN 2 - PEKAN 5 (PROSES AI JST)
# =====================================================================

import numpy as np
import math
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# --- PEKAN 2: DATA PREPROCESSING & SLIDING WINDOW ---
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_data = scaler.fit_transform(df[['Harga']])

def create_sliding_window(data, window_size):
    X_data, y_data = [], []
    for i in range(len(data) - window_size):
        X_data.append(data[i:(i + window_size), 0])
        y_data.append(data[i + window_size, 0])
    return np.array(X_data), np.array(y_data)

WINDOW_SIZE = 5
X, y = create_sliding_window(scaled_data, WINDOW_SIZE)

train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]


# --- PEKAN 3: DESAIN ARSITEKTUR JST ---
tf.random.set_seed(42)
model = Sequential([
    Dense(12, activation='relu', input_dim=WINDOW_SIZE), # Hidden Layer 1
    Dense(8, activation='relu'),                        # Hidden Layer 2
    Dense(1)                                             # Output Layer
])
model.compile(optimizer='adam', loss='mean_squared_error')


# --- PEKAN 4: PROSES TRAINING MODEL ---
print("=== Sedang Melatih AI (Training), Silakan Tunggu... ===")
history = model.fit(X_train, y_train, epochs=100, batch_size=32, validation_data=(X_test, y_test), verbose=0)
print("Proses Training Selesai!\n")

# Menampilkan grafik Loss Curve
plt.figure(figsize=(10, 4))
plt.plot(history.history['loss'], label='Train Loss (Error Belajar)', color='blue')
plt.plot(history.history['val_loss'], label='Validation Loss (Error Ujian)', color='orange')
plt.title('Grafik Loss Curve JST (Evaluasi Pekan 4)', fontsize=12)
plt.xlabel('Epochs', fontsize=10)
plt.ylabel('Loss Value (MSE)', fontsize=10)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()


# --- PEKAN 5: EVALUASI PREDIKSI MODEL ---
y_pred_scaled = model.predict(X_test)

# Proses Denormalisasi
y_test_actual = scaler.inverse_transform(y_test.reshape(-1, 1))
y_pred_actual = scaler.inverse_transform(y_pred_scaled)

# Menghitung nilai Performa Model berupa RMSE
rmse = math.sqrt(mean_squared_error(y_test_actual, y_pred_actual))
print("\n" + "="*35)
print(f" NILAI RMSE MODEL JST ANDA: {rmse:.2f} IDR")
print("="*35)

# Visualisasi Grafik Perbandingan Akhir
plt.figure(figsize=(12, 5))
plt.plot(y_test_actual, label='Harga Aktual (Asli Bursa)', color='blue', linewidth=2)
plt.plot(y_pred_actual, label='Harga Prediksi (Tebakan JST)', color='red', linestyle='--', linewidth=2)
plt.title('Perbandingan Nilai Aktual vs Hasil Prediksi Saham BBCA (Pekan 5)', fontsize=12)
plt.xlabel('Indeks Waktu (Data Testing)', fontsize=10)
plt.ylabel('Harga Saham (IDR)', fontsize=10)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
